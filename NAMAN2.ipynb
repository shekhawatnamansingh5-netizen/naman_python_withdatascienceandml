```python
import matplotlib.pyplot as plt
years=[1992,1994,1996,1998,2000,2002,2004,2006,2008,2010]
kolhi=[0,0,500,800,1100,1300,1500,1800,1900,2100]
sehwag=[0,300,800,1200,1500,1700,1600,1400,1800,0]
plt.plot(years,kolhi,'ro--',linewidth=10,label="kolhi run")
plt.plot(years,sehwag ,'g^:',linewidth=10,label="sehwag run")
plt.legend()
plt.grid(True)
plt.tight_layout()
plt.style.use("ggplot")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
years=[1992,1994,1996,1998,2000,2002,2004,2006,2008,2010]
sachin=[150,1000,1500,400,2000,300,200,1000,0,200]
kolhi=[0,0,500,800,1100,1300,1500,1800,1900,2100]
sehwag=[0,300,800,1200,1500,1700,1600,1400,1800,0]
x=np.arange(len(years))
width=0.1
plt.bar(x-width,sachin,width=width,label="sachin")
plt.bar(x,kolhi,width=width,label="kolhi")
plt.bar(x+width,sehwag,width=width,label="sehwag")
plt.xlabel("year")
plt.ylabel("runs")
plt.title("runs comparision")
plt.xticks(x,years)
plt.legend()
plt.tight_layout()
plt.show()




```
