```python
def word_frequencies(sentence):
    words=sentence.split()
    frequencies_dict={}
    for word in words:
        word_lower=word.lower()
        if word_lower in frequencies_dict:
            frequencies_dict[word_lower]+=1
        else:
            frequencies_dict[word_lower]=1
    return frequencies_dict
print( word_frequencies("my my name name is is naman naman"))
        
        
```

```python
def is_subset(main_set,subset_set):
    return subset_set.issubset(main_set)
set1={10,20,30,40}
set2={10,20}
set3={10,20,30}
print(f"is{set1} a subset of {set2}? {is_subset(set1,set2)}")
print(f"is{set3} a subset of {set2}? {is_subset(set3,set2)}")
```

```python
def removeduplicates(input_list):
 seen=set()
 result=[]
 for elements in input_list:
     if elements not in seen:
         
         seen.add(elements)
         result.append(elements)
 return result
list1=[1,1,2,2,3,3,4,5,6]
print(removeduplicates(list1))
```

```python
def evensqremoveodd(inputlist):
    list1=[]
    for i in inputlist:
        if i%2==0:
            i=i*i
            list1.append(i)
        
    return list1
lists=[1,2,3,4,5,6]
print(evensqremoveodd(lists))
```
