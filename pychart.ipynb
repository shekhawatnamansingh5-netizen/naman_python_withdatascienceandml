```python
import matplotlib.pyplot as plt
plt.style.use("ggplot")
players=['sachin','sehwag','kolhi','yuvraj']
runs=[18000,8000,12000,9500]
# Define colors before using them in plt.pie()
colors=['#ff9999','#6663ff','#99ff99','#ffcc99']  # Added missing '#' to the last color
plt.pie(runs,labels=players,startangle=140,colors=colors)
plt.title("carrer runs of ind players")
plt.show()
```

```python
plt.pie(runs,labels=players,colors=colors,wedgeprops={'edgecolor':'black','linewidth':2,"linestyle":'--'})
plt.show()
```
