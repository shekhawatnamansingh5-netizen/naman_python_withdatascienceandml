```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6, 7, 8, 9]
exam_scores = [40, 45, 50, 55, 60, 65, 75, 85, 90]

plt.scatter(study_hours, exam_scores)
plt.title('Study Hours vs Exam Scores')  # Fixed: Changed from 'exam hours' to 'Exam Scores'
plt.xlabel('Study Hours')
plt.ylabel('Exam Scores')  # Fixed: Changed from 'exam hours' to 'Exam Scores'
plt.grid(True)
plt.show()

# Note: If the error persists, restart your kernel to clear any variable conflicts
# that might be overriding matplotlib functions
```

```python
import numpy as np
scores_normalize=np.array(exam_scores)
plt.scatter(study_hours,exam_scores,c=scores_normalize,cmap='viridis')
plt.colorbar(label='score')
plt.title('Study Hours vs Exam Scores')
plt.xlabel('Study Hours')
plt.ylabel('Exam Scores')
plt.grid(True)
plt.show()
```

```python
plt.scatter(study_hours,exam_scores)
for i in range(len(study_hours)):
    plt.annotate(f's {i+1}',(study_hours[i],exam_scores[i]))
plt.title("scatter plot with annotation")
plt.xlabel("study hours")
plt.ylabel("exam scores")
plt.grid(True)
plt.show()
    
```
